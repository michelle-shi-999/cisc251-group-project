### Statistical Analysis of Sale Price
By looking at the shape and distribution of the graph, we can determine a few things:
- We can see that the graph displays a **right-skewed** distribution of houses and their sale prices. This implies that the model's accuracy may be influenced during training - especially if a regression model is used. Methods such as **Log Transformation** ( `np.log(SalePrice)` ) or **Square Root Transformation** may be used to modify the shape of the data to a more symmetric distribution and remove skewness.

- There is an **imbalance in target classes**. As the bulk of the sale prices fall on the left side of the graph, it means that the majority of the sale prices will fall under the "low" class, much fewer would be in the "middle" class and even fewer would be in the "high" class - so much so that some may even be considered outliers. Further outlier analysis is performed farther below on certain note-worthy features.

- Since `SalePrice` is a numerical feature (and that the relationships between certain features are quite linear as we will see farther below), it makes sense to use **Linear Regression** as our basis for predictive modelling.

### EDA of all features against sales price

```python
# The following code generates a scatter plot or box plot (depending on feature type) for each feature against sales price
target = 'SalePrice'
features = [col for col in df_train.columns if col != target and col != 'Id']

# Set up grid layout
n = len(features)
fig, axes = plt.subplots(nrows=(n + 2) // 3, ncols=3, figsize=(15, 5 * ((n + 2) // 3)))
axes = axes.flatten()

for i, feature in enumerate(features):
    if feature in categorical_cols:  # Categorical
        sns.boxplot(
            x=df_train[feature],
            y=df_train[target],
            hue=df_train[feature],
            ax=axes[i],
            palette="flare",
            legend=False
        )
        # Rotate labels if feature has more than 10 unique values
        if df_train[feature].nunique() > 10:
            axes[i].tick_params(axis='x', rotation=90)

    elif feature in numerical_cols:  # Numerical
        sns.scatterplot(x=df_train[feature],
                        y=df_train[target],
                        hue=df_train[feature],
                        ax=axes[i],
                        palette="flare",
                        legend=False
        )
    else:
        axes[i].set_visible(False)
        continue

# Hide any empty subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

#### Outlier Analysis of 4 above features:


While it may be enough to just find the outliers of each feature independent of the others, we believe that it is best to find patterns and see which house IDs appear in more than one feature as an outlier. This block of code outputs a table doing exactly that.

However, as we will primarily be using z-score analysis to mathematically compute the extreme outliers (by setting z = 4) and the top two features ( `OverallQual`, `YearBuilt` ) are categorical, we will only be looking at the bottom two features - `TotalBsmtSF` and `GrLivArea`.

Each outlier is first independently compared with their feature against SalePrice before being examined for common grounds.

```python
features = ['TotalBsmtSF', 'GrLivArea'] # The four features we picked out/took note of
outliers_dict = {}

# Find z-scores for each feature against SalePrice; if a z-score > 4, place it in the list of extreme outliers for the current feature.
# Then, place all the outlier lists into one dictionary to keep everything together.
for f in features:
    z = np.abs(stats.zscore(df_train[[f, 'SalePrice']]))
    outliers = df_train[(z > 4).any(axis=1)]['Id'].tolist()
    outliers_dict[f] = set(outliers)

# Find the houses that count as outliers in both features and place their IDs in a list
common_outliers = outliers_dict['TotalBsmtSF'].intersection(outliers_dict['GrLivArea'])

# Create a table to output all common outliers
outliers_table = df_train[df_train['Id'].isin(common_outliers)][['Id', 'TotalBsmtSF', 'GrLivArea', 'SalePrice']] 

 # Finally, print out the resulting table
print("Common Outliers Between TotalBsmtSF and GrLivArea\n")
print(outliers_table)
```

Looking at the resulting outlier analysis table, it appears that there are 11 house IDs that are labelled as extreme outliers in both the `TotalBsmtSF` and `GrLivArea` features. They consist of houses that are generally large yet super cheap (e.g. the house with `Id=1299`), and houses that are on the smaller size and yet very expensive (e.g. the houses with `Id = 692, 1183`).

When performing an EDA on any dataset, it is important to label these outliers, as they may hold a lot of influence on the model's training and thus its output and accuracy. Because we originally picked these features due to their high linear correlation with SalePrice, such outliers plus the skewed distribution of the data can greatly affect the model's predicted linear relationships.

Therefore, we believe it is best to simply remove these outliers, since they are complete anomalies and we have enough data to allow the loss of just a few data points. Doing so will also refine the general linear relationship and, in turn, the predictive model as well.

# Challenges and Strategies

During the analysis of the Ames housing dataset, several key challenges were identified that could impact the performance of a linear regression model. Below are the main issues we recognized and how they will be handled.

### Missing Values
Several features contained missing data, often because the property simply didn’t have that feature (e.g., no pool, no garage, or no alley). In those cases, the dataset already has them labelled with `"N/A"` to indicate absence. For numerical variables such as `LotFrontage`, imputation can be applied using the median value of nearby homes to maintain consistency. Features with a large amount of missing data and low importance (like `PoolQC`) may be dropped.

### Outliers and Skewed Data
Exploratory analysis showed a few extreme outliers, especially in `GrLivArea` and `TotalBsmtSF`. These can strongly influence regression results, so they can be checked manually and removed if they are unrealistic. The target variable `SalePrice` was right-skewed, so applying a log transformation (`log(SalePrice)`) could help make it more normally distributed and improves model performance.

### High Dimensionality and Categorical Encoding
The dataset has over 70 features, many of which are categorical. Ordinal variables (such as `OverallQual`, `ExterQual`, and `KitchenQual`) are mapped to numerical scales, while nominal features can be one-hot encoded. To reduce dimensionality and avoid overfitting, low-variance or redundant features can be dropped after checking correlations.

### Multicollinearity
Some features in the dataset are closely related to each other (for example, `GrLivArea`, `1stFlrSF`, and `TotalBsmtSF` all describe the size of the house). This can cause overlap in the information the model learns. To reduce this, correlations between numerical features can be checked, and if two features are highly related, one may be removed or combined to make the model more stable.

### Feature Scaling
Numerical features vary widely in scale, so standardization with z-scores can be used to make sure all variables contribute equally to the model. This also helps the regression model converge more efficiently and makes coefficients easier to interpret.

### Rare Categories
A few categorical values appeared only a handful of times, these can be grouped into an `"Other"` category to simplify the model and improve generalization.

---

**Summary:**  
By identifying missing data, spotting outliers and skewed distributions, and planning appropriate preprocessing steps such as encoding categorical variables and scaling numeric features, the dataset will be prepared for a robust linear regression model. These strategies help ensure the model is accurate, stable, and generalizable once the data is cleaned and preprocessed.
