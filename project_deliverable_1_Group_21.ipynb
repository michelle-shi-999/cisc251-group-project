### EDA of all features against sales price

```python
# The following code generates a scatter plot or box plot (depending on feature type) for each feature against sales price
target = 'SalePrice'
features = [col for col in df_train.columns if col != target and col != 'Id']

# Set up grid layout
n = len(features)
fig, axes = plt.subplots(nrows=(n + 2) // 3, ncols=3, figsize=(15, 5 * ((n + 2) // 3)))
axes = axes.flatten()

for i, feature in enumerate(features):
    if feature in categorical_cols:  # Categorical
        sns.boxplot(
            x=df_train[feature],
            y=df_train[target],
            hue=df_train[feature],
            ax=axes[i],
            palette="flare",
            legend=False
        )
        # Rotate labels if feature has more than 10 unique values
        if df_train[feature].nunique() > 10:
            axes[i].tick_params(axis='x', rotation=90)

    elif feature in numerical_cols:  # Numerical
        sns.scatterplot(x=df_train[feature],
                        y=df_train[target],
                        hue=df_train[feature],
                        ax=axes[i],
                        palette="flare",
                        legend=False
        )
    else:
        axes[i].set_visible(False)
        continue

# Hide any empty subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

#### There are lot's of features here, but not all show clear positive or negative linear correlation.

### Frequency of Missing Data

```python
missing_table = (
    df_train.isnull()
      .sum()
      .reset_index()
      .rename(columns={'index': 'Feature', 0: 'MissingValues'})
)
missing_table['MissingPercent'] = 100 * missing_table['MissingValues'] / len(df_train)
missing_table = missing_table.sort_values(by='MissingValues', ascending=False)
missing_table = missing_table[missing_table['MissingValues'] > 0]
print(missing_table)
```

Looking at the missing values above we can see they are explained in the data_description.txt file. NA is described in this file for almost all the above cases, as that property is simply lacking the relevant feature.

For example, if we look at all 4 features related to garages, we can see that 81 houses are missing data for each feature. This is because 81 properties don't have garages of any kind.
